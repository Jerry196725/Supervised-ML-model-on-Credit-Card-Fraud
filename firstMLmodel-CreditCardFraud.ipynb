Fine tuning my model to get better performance.

```python
from sklearn.metrics import mean_absolute_error
from sklearn.tree import DecisionTreeRegressor

def get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y):
    CCFraud_model2 = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes, random_state=0)
    CCFraud_model2.fit(train_X, train_y)
    pred_val = CCFraud_model2.predict(val_X)
    mae = mean_absolute_error(val_y, pred_val)
    return(mae)

# Comparing MAE with differing values of max leaf nodes
for max_leaf_nodes in [5, 50, 500, 5000]:
    my_mae = get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y)
    print("Max leaf nodes: %d  \t\t Mean Absolute Error: %d" %(max_leaf_nodes, my_mae))
```
